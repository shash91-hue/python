```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
```

```python
mpg = sns.load_dataset("mpg")
```

```python
mpg.head(2)
```

```python
#Find cars with horsepower greater than 150.
cars_hp = mpg[mpg['horsepower']>150][['name','horsepower']].head(10)
cars_hp
```

```python
# Find the average MPG for each cylinder category.
avg_mpg_c = mpg.groupby('cylinders')['mpg'].mean()
avg_mpg_c
```

```python
# Replace missing horsepower values with the median horsepower.
miss_hp = mpg['horsepower'].median()
#miss_hp
mpg['horsepower'].fillna(miss_hp, inplace = True)
```

```python
# Find cars with MPG between 20 and 30.
cars_mpg = mpg[(mpg['mpg']>=20) & (mpg['mpg']<=30)].reset_index()
cars_mpg
```

```python
mpg.columns
```

```python
# Find cars from USA with MPG greater than 25
usa_cars_25 = mpg[(mpg["origin"] == 'usa') & (mpg['mpg']>25)][['name','origin','mpg']].head(10).reset_index()
usa_cars_25
```
